根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$

4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$

### 问题1 解答

#### 1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
**求解步骤**：
- 采用**主定理**求解，递归式符合主定理标准形式 $T(n)=aT(n/b)+f(n)$，其中 $a=1, b=4, f(n)=1$
- 计算临界指数：$\log_b a = \log_4 1 = 0$，因此 $n^{\log_b a} = n^0 = 1$
- 对比$f(n)$与临界项：$f(n)=1 = \Theta(n^{\log_b a})$，符合主定理**情况2**
- 结论：$T(n) = \Theta(log n)$，紧渐近上界为 $O(log n)$

#### 2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$
**求解步骤**：
- 主定理标准形式，$a=3, b=3, f(n)=n^2$
- 临界指数：$\log_b a = \log_3 3 = 1$，$n^{\log_b a} = n^1 = n$
- 对比$f(n)$与临界项：$f(n)=n^2 = \Omega(n^{1+\varepsilon})$，取$\varepsilon=1>0$，符合主定理**情况3**
- 验证正则条件：$a f(n/b) = 3*(n/3)^2 = n^2/3 \leq c n^2$，取$c=1/3<1$，条件成立
- 结论：$T(n) = \Theta(n^2)$，紧渐近上界为 $O(n^2)$

#### 3. $T(1)=1;T(n)=T(2n/3)+1$ for $n>1$
**求解步骤**：
- 主定理标准形式，$a=1, b=3/2, f(n)=1$
- 临界指数：$\log_b a = \log_{3/2} 1 = 0$，$n^{\log_b a} = 1$
- 对比$f(n)$与临界项：$f(n)=1 = \Theta(n^{\log_b a})$，符合主定理**情况2**
- 结论：$T(n) = \Theta(log n)$，紧渐近上界为 $O(log n)$

#### 4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$
**求解步骤**：
- 主定理标准形式，$a=5, b=4, f(n)=n$
- 临界指数：$\log_b a = \log_4 5 \approx 1.16$，$n^{\log_b a} = n^{\log_4 5}$
- 对比$f(n)$与临界项：$f(n)=n = O(n^{\log_4 5 - \varepsilon})$，取$\varepsilon\approx0.16>0$，符合主定理**情况1**
- 结论：$T(n) = \Theta(n^{\log_4 5})$，紧渐近上界为 $O(n^{\log_4 5})$

#### 5. $T(n)=1 \ for\ n \le 2 ; T(n)=T(\sqrt{n})+1 \ for \ n>2$
**求解步骤**：
- 采用**变量替换法**简化递归式：令 $m = log_2 n$，即 $n=2^m$，代入递归式得：
  $T(2^m) = T(2^{m/2}) + 1$
- 令 $S(m) = T(2^m)$，则递归式转化为：$S(m) = S(m/2) + 1$，边界条件 $S(1)=T(2)=1$
- 对$S(m)$用主定理求解：$a=1,b=2,f(n)=1$，符合情况2，得 $S(m) = \Theta(log m)$
- 回代变量：$T(n) = S(log n) = \Theta(log log n)$，紧渐近上界为 $O(log log n)$

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

### 问题3 解答

#### 算法设计与分析
题目要求只能一次买入、一次卖出，且卖出必须在买入之后，核心目标是找到最大利润及对应日期。
- **暴力解法**：双重循环遍历所有买入卖出组合，计算利润取最大值，时间复杂度$O(n^2)$。具体实现：枚举买入日i和卖出日j（i<j），计算从i到j的累计收益，记录最大值。
- 时间复杂度$O(n^2)$，空间复杂度$O(1)$。

注：题目示例中`profits=[3,2,1,-7,5,2,-1,3,-1]`，数组索引从0开始，第4位（索引4）是5，对应第5天，第7位（索引7）是3，对应第8天，利润计算为从买入日到卖出日的收益累加。

````python
# 暴力法实现（对应第1小问）
def max_profit_bruteforce(profits):
    n = len(profits)
    max_profit = 0
    buy_day = -1
    sell_day = -1
    for i in range(n):
        for j in range(i+1, n):
            # 计算从i到j的累计收益
            profit = sum(profits[i:j+1])
            if profit > max_profit:
                max_profit = profit
                buy_day = i + 1
                sell_day = j + 1
    return max_profit, buy_day, sell_day

# 测试题目示例
profits_example = [3,2,1,-7,5,2,-1,3,-1]
profit, buy, sell = max_profit_bruteforce(profits_example)
print(f"最大利润: {profit}")
print(f"买入日期: 第{buy}天")
print(f"卖出日期: 第{sell}天")
print("暴力法时间复杂度: O(n^2)")
````

````python
# 最大利润: 9
# 买入日期: 第5天
# 卖出日期: 第8天
# 暴力法时间复杂度: O(n^2)
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

### 问题4 解答

#### 1. O(logn)时间复杂度找一个局部最小值
**算法设计思路**：
题目给定了边界条件$A[1]>A[2]$、$A[n-1]<A[n]$，保证数组中一定存在局部最小值，因此可以采用**二分查找法**实现O(logn)的时间复杂度：
1.  取数组中间位置`mid`，检查`mid`是否为局部最小值：若`A[mid] < A[mid-1]`且`A[mid] < A[mid+1]`，直接返回该值。
2.  若`mid`不是局部最小值，比较`mid`左右邻居：
    - 若`A[mid-1] < A[mid]`：左半部分一定存在局部最小值（因为左边界满足$A[1]>A[2]$，左半部分仍符合边界条件），递归/迭代查找左半部分。
    - 否则：右半部分一定存在局部最小值（右边界满足$A[n-1]<A[n]$），递归/迭代查找右半部分。
3.  每次查找将数组规模缩小一半，因此时间复杂度为$O(logn)$，空间复杂度可优化为$O(1)$（迭代实现）。

````python
# 1. O(logn)时间复杂度找一个局部最小值（迭代实现，无递归栈开销）
def find_one_local_min(A):
    n = len(A)
    if n < 3:
        return None  # 不符合题目边界条件，无局部最小值
    left = 0
    right = n - 1
    
    while left <= right:
        mid = (left + right) // 2
        # 边界处理：题目保证A[1]>A[2]、A[n-1]<A[n]，mid不会在边界
        if mid == 0 or mid == n-1:
            break
        # 找到局部最小值
        if A[mid] < A[mid-1] and A[mid] < A[mid+1]:
            return A[mid]
        # 左半部分有更小值，查找左半
        elif A[mid-1] < A[mid]:
            right = mid - 1
        # 右半部分有更小值，查找右半
        else:
            left = mid + 1
    return None

# 测试题目示例
A_example = [9, 3, 7, 2, 1, 4, 5]
one_min = find_one_local_min(A_example)

print(f"找到的一个局部最小值: {one_min}")
print("找单个局部最小值时间复杂度: O(logn)")
````

````python
# 找到的一个局部最小值: 1
# 找单个局部最小值时间复杂度: O(logn)
````

## 问题 5

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

### 问题5 解答

#### 1. 排序法实现
**算法设计**：选择工程上高效的Timsort排序算法（Python内置sort方法的底层实现），对数组进行全排序后，取前k个元素即为最小的k个数。
- 时间复杂度：排序的时间复杂度为$O(n log n)$，取前k个元素为$O(k)$，总时间复杂度$O(n log n)$。
- 优点：实现简单，代码简洁，适合n规模不大的场景。

#### 2. 插入法建堆实现
**算法设计**：
1.  初始化一个空的小顶堆，遍历数组所有元素，逐个插入堆中（插入法建堆）。
2.  连续执行k次堆的Extract-min操作（取出堆顶最小值，然后调整堆），得到的k个值即为按顺序排列的最小k个数。
- 建堆时间复杂度分析：每次堆插入操作的时间复杂度为$O(log n)$，n次插入总时间复杂度为$O(n log n)$；k次Extract-min操作每次为$O(log n)$，总时间复杂度$O(k log n)$；整体时间复杂度为$O(n log n + k log n) = O(n log n)$。

````python
import heapq

# 1. 排序法实现
def top_k_min_sort(arr, k):
    if k <= 0 or k > len(arr):
        return []
    # 全排序
    sorted_arr = sorted(arr)
    # 取前k个
    return sorted_arr[:k]

# 2. 插入法建堆实现
def top_k_min_heap_insert(arr, k):
    if k <= 0 or k > len(arr):
        return []
    heap = []
    # 插入法建堆：逐个插入元素到空堆
    for num in arr:
        heapq.heappush(heap, num)
    # k次extract-min，取出前k小
    top_k = [heapq.heappop(heap) for _ in range(k)]
    return top_k

# 测试题目示例
arr_example = [5,4,3,2,6,1,88,33,22,107]
k_example = 3

print(f"排序法结果: {top_k_min_sort(arr_example, k_example)}")
print(f"排序法时间复杂度: O(n log n)")

print(f"\n插入法建堆结果: {top_k_min_heap_insert(arr_example, k_example)}")
print(f"插入法建堆时间复杂度: O(n log n)")
````

````python
# 排序法结果: [1, 2, 3]
# 排序法时间复杂度: O(n log n)
#
# 插入法建堆结果: [1, 2, 3]
# 插入法建堆时间复杂度: O(n log n)
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

### 问题6 解答

#### 算法设计思路
星球碰撞的核心逻辑是：**只有向左运动的负数星球，才会和右侧向左、右侧向右的星球碰撞**，本质是相邻的「正数（向右）+负数（向左）」才会发生碰撞，因此采用**栈**来模拟碰撞过程是最优方案：
1.  初始化一个空栈，用于存储最终存续的星球。
2.  遍历数组中的每个星球：
    - 若当前星球是**正数（向右运动）**：不会和栈中已有的星球发生碰撞（栈中星球在左侧，同向不会撞，向左的星球会远离），直接入栈。
    - 若当前星球是**负数（向左运动）**：需要循环和栈顶的星球碰撞，直到无法碰撞为止：
      1.  若栈为空，或栈顶是负数（向左运动）：同向不会碰撞，当前星球入栈，结束循环。
      2.  若栈顶是正数（向右运动）：发生碰撞，比较两者绝对值：
          - 当前星球绝对值更大：栈顶星球消失（弹出栈），继续循环碰撞。
          - 两者绝对值相等：栈顶星球弹出，当前星球消失，结束循环。
          - 当前星球绝对值更小：当前星球消失，结束循环。
3.  遍历完成后，栈中剩余的元素就是最终存续的星球。

#### 复杂度分析
- 时间复杂度：$O(n)$。每个星球最多入栈1次、出栈1次，总操作次数为线性级别。
- 空间复杂度：$O(n)$。最坏情况下所有星球都不会碰撞，栈需要存储全部n个元素。

````python
# 星球碰撞算法实现
def planet_collision(A):
    stack = []
    for planet in A:
        # 当前星球向左运动（负数），需要处理碰撞
        while stack and planet < 0 < stack[-1]:
            # 比较质量绝对值
            if abs(planet) > stack[-1]:
                # 当前星球质量更大，栈顶星球消失
                stack.pop()
            elif abs(planet) == stack[-1]:
                # 质量相等，栈顶弹出，当前星球消失
                stack.pop()
                planet = 0
                break
            else:
                # 当前星球质量更小，消失
                planet = 0
                break
        # 若当前星球未消失，入栈
        if planet != 0:
            stack.append(planet)
    return stack

# 测试题目示例
test1 = [-3,-6,2,8,5,-8,9,-2,1]
test2 = [23,-8,9,-3,-7,9,-23,22]

print(f"示例1输入: {test1}")
print(f"示例1输出: {planet_collision(test1)}")
print(f"\n示例2输入: {test2}")
print(f"示例2输出: {planet_collision(test2)}")
print(f"\n算法时间复杂度: O(n)，空间复杂度: O(n)")
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

### 问题7 解答

#### 算法设计思路
这是经典的**0-1背包问题**，核心特点是每个物品只能选或不选，不能分割，目标是在承重限制下最大化总价值。采用**动态规划（DP）** 是最优解法：
1.  **状态定义**：定义二维数组`dp[i][w]`表示前`i`个物品，在总承重为`w`时的最大价值。
2.  **状态转移方程**：
    - 若不选第`i`个物品：`dp[i][w] = dp[i-1][w]`
    - 若选第`i`个物品（需满足承重足够）：`dp[i][w] = dp[i-1][w - weight_i] + value_i`
    - 取两者最大值：`dp[i][w] = max(dp[i-1][w], dp[i-1][w - weight_i] + value_i)`
3.  **边界条件**：`dp[0][w] = 0`（无物品时价值为0），`dp[i][0] = 0`（承重为0时价值为0）。
4.  最终结果：`dp[n][W]`，其中`n`为物品总数，`W`为最大承重。

#### 复杂度分析
- 时间复杂度：$O(n \times W)$，其中`n`是物品数量，`W`是最大承重。需要填充一个`n×W`的DP表。
- 空间复杂度：$O(n \times W)$，可优化为$O(W)$（使用一维DP数组）。

#### 本题手动推导验证
物品：[(5,10), (4,40), (6,30), (3,50)]，承重W=9
- 最优选择：选第2个（重量4，价值40）和第4个（重量3，价值50），总重量7≤9，总价值90。

````python
# 0-1背包问题动态规划实现
def knapsack_01(items, max_weight):
    n = len(items)
    # dp[i][w]：前i个物品，承重w时的最大价值
    dp = [[0] * (max_weight + 1) for _ in range(n + 1)]
    
    for i in range(1, n + 1):
        weight_i, value_i = items[i-1]
        for w in range(1, max_weight + 1):
            # 不选第i个物品
            not_take = dp[i-1][w]
            # 选第i个物品（如果承重足够）
            take = dp[i-1][w - weight_i] + value_i if w >= weight_i else 0
            # 取最大值
            dp[i][w] = max(not_take, take)
    
    return dp[n][max_weight]

# 测试题目示例
marbles = [(5, 10), (4, 40), (6, 30), (3, 50)]
max_load = 9
max_value = knapsack_01(marbles, max_load)

print(f"大理石列表（重量，价值）: {marbles}")
print(f"小推车最大承重: {max_load}")
print(f"最大价值为: {max_value}")
print(f"算法时间复杂度: O(n*W) = O(4*9) = O(36)")
print(f"算法空间复杂度: O(n*W)")
````

````python
# 大理石列表（重量，价值）: [(5, 10), (4, 40), (6, 30), (3, 50)]
# 小推车最大承重: 9
# 最大价值为: 90
# 算法时间复杂度: O(n*W) = O(4*9) = O(36)
# 算法空间复杂度: O(n*W)
````

输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]，

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

### 问题8 解答

#### 算法设计思路
二叉搜索树（BST）的核心性质是：**中序遍历的结果是严格递增的序列**。题目中恰好有两个节点被交换，因此中序遍历序列中会出现**两处逆序对**，我们可以利用这一特性恢复BST：
1.  **中序遍历BST**：记录遍历过程中的节点。
2.  **寻找逆序节点**：
    - 定义`prev`指针记录前一个访问的节点。
    - 第一次遇到`prev.val > current.val`时，`prev`是第一个错误节点。
    - 第二次遇到`prev.val > current.val`时，`current`是第二个错误节点。
3.  **交换值**：交换找到的两个错误节点的值，不改变树结构。

为了优化空间，可采用**Morris中序遍历**，将空间复杂度降为$O(1)$。

#### 复杂度分析
- 时间复杂度：$O(n)$，需遍历所有节点一次。
- 空间复杂度：$O(h)$（递归/栈实现，h为树高），最优为$O(1)$（Morris遍历）。

````python
# 定义二叉树节点（仅定义一次，避免重复）
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# 恢复二叉搜索树：递归中序遍历实现（易读，空间O(h)）
class Solution:
    def __init__(self):
        self.first = None  # 第一个错误节点
        self.second = None # 第二个错误节点
        self.prev = None    # 前一个节点
    
    def recoverTree(self, root):
        """
        Do not return anything, modify root in-place instead.
        """
        # 中序遍历找错误节点
        self.inorder(root)
        # 交换两个错误节点的值
        if self.first and self.second:
            self.first.val, self.second.val = self.second.val, self.first.val
    
    def inorder(self, node):
        if not node:
            return
        # 遍历左子树
        self.inorder(node.left)
        # 处理当前节点：找逆序对
        if self.prev and self.prev.val > node.val:
            if not self.first:
                self.first = self.prev  # 第一个错误节点
            self.second = node         # 第二个错误节点
        self.prev = node
        # 遍历右子树
        self.inorder(node.right)

# 辅助函数：层序遍历打印树（用于验证）
def print_tree_level_order(root):
    if not root:
        return []
    result = []
    from collections import deque
    q = deque([root])
    while q:
        node = q.popleft()
        if node:
            result.append(node.val)
            q.append(node.left)
            q.append(node.right)
        else:
            result.append(None)
    # 去掉末尾的None
    while result and result[-1] is None:
        result.pop()
    return result

# 测试示例1: [1,3,null,null,2]
root1 = TreeNode(1)
root1.left = TreeNode(3)
root1.left.right = TreeNode(2)
print(f"示例1输入: {print_tree_level_order(root1)}")
sol = Solution()
sol.recoverTree(root1)
print(f"示例1输出: {print_tree_level_order(root1)}")

# 测试示例2: [3,1,4,null,null,2]
root2 = TreeNode(3)
root2.left = TreeNode(1)
root2.right = TreeNode(4)
root2.right.left = TreeNode(2)
print(f"\n示例2输入: {print_tree_level_order(root2)}")
sol2 = Solution()
sol2.recoverTree(root2)
print(f"示例2输出: {print_tree_level_order(root2)}")
print(f"\n算法时间复杂度: O(n)")
print(f"算法空间复杂度: O(h)，h为树高")
````

````python
# 示例1输入: [1, 3, None, None, 2]
# 示例1输出: [3, 1, None, None, 2]
#
# 示例2输入: [3, 1, 4, None, None, 2]
# 示例2输出: [2, 1, 4, None, None, 3]
#
# 算法时间复杂度: O(n)
# 算法空间复杂度: O(h)，h为树高
````

给你二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

### 问题10 解答

#### 算法设计思路
这是典型的**二叉树深度优先搜索（DFS）** 问题，核心是遍历每条从根到叶子的路径，并累加路径代表的数字：
1.  **递归定义**：定义辅助函数`dfs(node, current_num)`，其中`current_num`表示从根到当前节点的路径数字。
2.  **路径数字更新**：每深入一层，`current_num = current_num * 10 + node.val`（例如路径1->2，更新为1*10+2=12）。
3.  **终止条件**：当到达叶子节点（`node.left is None and node.right is None`）时，将`current_num`累加到总结果中。
4.  **递归过程**：递归遍历左子树和右子树。

#### 复杂度分析
- 时间复杂度：$O(n)$，需遍历每个节点恰好一次。
- 空间复杂度：$O(h)$，其中`h`为树的高度，由递归栈占用；最坏情况（树退化为链表）为$O(n)$。

````python
# 二叉树路径数字之和：DFS递归实现
class SolutionSumNumbers:
    def sumNumbers(self, root):
        self.total_sum = 0
        self.dfs(root, 0)
        return self.total_sum
    
    def dfs(self, node, current_num):
        if not node:
            return
        # 更新当前路径的数字
        current_num = current_num * 10 + node.val
        # 到达叶子节点，累加到总和
        if not node.left and not node.right:
            self.total_sum += current_num
            return
        # 递归遍历左右子树
        self.dfs(node.left, current_num)
        self.dfs(node.right, current_num)

# 测试示例1: [1,2,3]
root1 = TreeNode(1)
root1.left = TreeNode(2)
root1.right = TreeNode(3)
sol1 = SolutionSumNumbers()
print(f"示例1输入: [1,2,3]")
print(f"示例1输出: {sol1.sumNumbers(root1)}")

# 测试示例2: [4,9,0,5,1]
root2 = TreeNode(4)
root2.left = TreeNode(9)
root2.right = TreeNode(0)
root2.left.left = TreeNode(5)
root2.left.right = TreeNode(1)
sol2 = SolutionSumNumbers()
print(f"\n示例2输入: [4,9,0,5,1]")
print(f"示例2输出: {sol2.sumNumbers(root2)}")
print(f"\n算法时间复杂度: O(n)")
print(f"算法空间复杂度: O(h)，h为树高")
````
